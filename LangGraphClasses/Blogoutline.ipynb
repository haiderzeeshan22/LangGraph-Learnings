```python
from langgraph.graph import StateGraph, START, END
from typing import TypedDict
from langchain_openai import ChatOpenAI
from dotenv import load_dotenv
load_dotenv()
model = ChatOpenAI(model="gpt-4o-mini")
```

We give topic to the LLM to write a blog and LLM first create an outline for that topic and then will write/generate blog as per the outline

```python
class BlogState(TypedDict):
    title : str
    outline: str
    content : str

```

```python
def create_outline(state:BlogState)-> BlogState:
    title = state['title']
    prompt = f"Generate an outline for a blog on the topic - {title} "
    outline = model.invoke(prompt).content
    state['outline'] = outline

    return state
```

```python
def create_blog(state:BlogState)-> BlogState:
    title = state['title']
    outline = state['outline']

    prompt = f"write a detailed blog on the topic {title} using the following \n {outline}"

    content = model.invoke(prompt).content

    state['content'] = content

    return state
```

```python
graph = StateGraph(BlogState)

graph.add_node("create_outline", create_outline)
graph.add_node("create_blog", create_blog)

graph.add_edge(START, create_outline)  # create_outline and below create_blog must be str 
graph.add_edge('create_outline', create_blog)
graph.add_edge('create_blog', END)

workflow = graph.compile()
```

```python
graph = StateGraph(BlogState)

graph.add_node("create_outline", create_outline)
graph.add_node("create_blog", create_blog)

graph.add_edge(START, "create_outline")
graph.add_edge('create_outline', "create_blog")
graph.add_edge('create_blog', END)

workflow = graph.compile()
```

```python
initail_state = {"title": "Rise of AI in Pakistan?"}
final_state = workflow.invoke(initail_state)
print(final_state)
```
